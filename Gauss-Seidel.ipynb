```python
a = []
b = []

for i in range(3):
    a.append(list(map(float, input().split())))

b = list(map(float, input().split()))

x1 = x2 = x3 = 0

for i in range(10):
    x1 = (b[0] - x2 - x3) / a[0][0]
    x2 = (b[1] - a[1][0]*x1 - x3) / a[1][1]
    x3 = (b[2] - a[2][0]*x1 - a[2][1]*x2) / a[2][2]

    print(x1, x2, x3)

print(x1, x2, x3)
```
