```python
a = []

for i in range(3):
    a.append(list(map(float, input().split())))

for i in range(3):
    for j in range(i+1, 3):
        ratio = a[j][i] / a[i][i]
        for k in range(4):
            a[j][k] -= ratio * a[i][k]

x = [0,0,0]

for i in range(2, -1, -1):
    x[i] = a[i][3]
    for j in range(i+1, 3):
        x[i] -= a[i][j]*x[j]
    x[i] /= a[i][i]

print(x[0], x[1], x[2])
```
