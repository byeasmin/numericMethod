```python
def det3(a):
    return (
        a[0][0]*(a[1][1]*a[2][2] - a[1][2]*a[2][1])
      - a[0][1]*(a[1][0]*a[2][2] - a[1][2]*a[2][0])
      + a[0][2]*(a[1][0]*a[2][1] - a[1][1]*a[2][0])
    )

A = []
for i in range(3):
    A.append(list(map(float, input().split())))

B = list(map(float, input().split()))

A1 = [[B[0], A[0][1], A[0][2]],
      [B[1], A[1][1], A[1][2]],
      [B[2], A[2][1], A[2][2]]]

A2 = [[A[0][0], B[0], A[0][2]],
      [A[1][0], B[1], A[1][2]],
      [A[2][0], B[2], A[2][2]]]

A3 = [[A[0][0], A[0][1], B[0]],
      [A[1][0], A[1][1], B[1]],
      [A[2][0], A[2][1], B[2]]]

detA = det3(A)

print(det3(A1)/detA, det3(A2)/detA, det3(A3)/detA)
```
