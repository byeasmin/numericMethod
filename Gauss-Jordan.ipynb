```python
a = []

for i in range(3):
    a.append(list(map(float, input().split())))

for i in range(3):
    pivot = a[i][i]
    for j in range(4):
        a[i][j] /= pivot
    for k in range(3):
        if k != i:
            ratio = a[k][i]
            for j in range(4):
                a[k][j] -= ratio * a[i][j]

print(a[0][3], a[1][3], a[2][3])
```
