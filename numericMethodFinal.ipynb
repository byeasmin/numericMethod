```python
# Gauss elimination

a = []
for i in range(3):
    a.append(list(map(float, input().split())))
for i in range(3):
    for j in range(i+1, 3):
        ratio = a[j][i] / a[i][i]
        for k in range(4):
            a[j][k] -= ratio * a[i][k]
x = [0,0,0]
for i in range(2, -1, -1):
    x[i] = a[i][3]
    for j in range(i+1, 3):
        x[i] -= a[i][j]*x[j]
    x[i] /= a[i][i]
print(x[0], x[1], x[2])




# Gauss jordan

a = []
for i in range(3):
    a.append(list(map(float, input().split())))
for i in range(3):
    pivot = a[i][i]
    for j in range(4):
        a[i][j] /= pivot
    for k in range(3):
        if k != i:
            ratio = a[k][i]
            for j in range(4):
                a[k][j] -= ratio * a[i][j]
print(a[0][3], a[1][3], a[2][3])




# Jacobi

def jacobi(A, b, n, x, max_iter, tol):
    for it in range(max_iter):
        x_new = [0.0] * n
        for i in range(n):
            s = 0.0
            for j in range(n):
                if j != i:
                    s += A[i][j] * x[j]
            x_new[i] = (b[i] - s) / A[i][i]
        error = max(abs(x_new[i] - x[i]) for i in range(n))
        print("Iteration", it + 1, ":", x_new)
        if error < tol:
            return x_new
        x = x_new[:]
    return x
n = int(input())
A = []
b = []
for i in range(n):
    r = list(map(float, input().split()))
    A.append(r[:n])
    b.append(r[n])
x = [0.0] * n
max_iter = int(input())
tol = float(input())
solution = jacobi(A, b, n, x, max_iter, tol)
print("Final Answer:")
for i in range(n):
    print(solution[i])






# Inverse matrix

A = []
for i in range(3):
    A.append(list(map(float, input().split())))
B = list(map(float, input().split()))
def det(m):
    return (
        m[0][0] * (m[1][1] * m[2][2] - m[1][2] * m[2][1])
        - m[0][1] * (m[1][0] * m[2][2] - m[1][2] * m[2][0])
        + m[0][2] * (m[1][0] * m[2][1] - m[1][1] * m[2][0])
    )
def cof(m):
    c = [[0] * 3 for i in range(3)]
    c[0][0] = (m[1][1] * m[2][2] - m[1][2] * m[2][1])
    c[0][1] = -(m[1][0] * m[2][2] - m[1][2] * m[2][0])
    c[0][2] = (m[1][0] * m[2][1] - m[1][1] * m[2][0])

    c[1][0] = -(m[0][1] * m[2][2] - m[0][2] * m[2][1])
    c[1][1] = (m[0][0] * m[2][2] - m[0][2] * m[2][0])
    c[1][2] = -(m[0][0] * m[2][1] - m[0][1] * m[2][0])

    c[2][0] = (m[0][1] * m[1][2] - m[0][2] * m[1][1])
    c[2][1] = -(m[0][0] * m[1][2] - m[0][2] * m[1][0])
    c[2][2] = (m[0][0] * m[1][1] - m[0][1] * m[1][0])
    return c
def tr(m):
    t = []
    for i in range(3):
        r = []
        for j in range(3):
            r.append(m[j][i])
        t.append(r)
    return t
def inv(m):
    d = det(m)
    if d == 0:
        return None
    adj = tr(cof(m))
    x = []
    for i in range(3):
        r = []
        for j in range(3):
            r.append(adj[i][j] / d)
        x.append(r)
    return x
def mul(a, b):
    x = []
    for i in range(3):
        s = 0
        for j in range(3):
            s = s + a[i][j] * b[j]
        x.append(s)
    return x
A_inv = inv(A)
if A_inv == None:
    print("No Inverse")
else:
    X = mul(A_inv, B)
    print("x =", X[0])
    print("y =", X[1])
    print("z =", X[2])






# 1st order & 2nd order derivatives

n = int(input("Enter number of data points: "))
x = []
y = []
for i in range(n):
    xi = float(input(f"Enter x[{i+1}]: "))
    yi = float(input(f"Enter y[{i+1}]: "))
    x.append(xi)
    y.append(yi)
h = x[1] - x[0]
d1 = []
for i in range(n - 1):
    d1.append(y[i + 1] - y[i])
d2 = []
for i in range(len(d1) - 1):
    d2.append(d1[i + 1] - d1[i])
d3 = []
for i in range(len(d2) - 1):
    d3.append(d2[i + 1] - d2[i])
d4 = []
for i in range(len(d3) - 1):
    d4.append(d3[i + 1] - d3[i])
first_derivative = (d1[0] - d2[0]/2 + d3[0]/3 - d4[0]/4) / h
second_derivative = (d2[0] - d3[0] + (11*d4[0])/12) / (h*h)
print(f"\nFirst Derivative at x = {x[0]} is {first_derivative:.2f}")
print(f"Second Derivative at x = {x[0]} is {second_derivative:.2f}")




# Simpon 1/3

def factorial(n):
    fact = 1
    for i in range(1, n + 1):
        fact *= i
    return fact
def sin(x):
    result = 0
    sign = 1
    for i in range(10):
        power = 2 * i + 1
        result += sign * (x ** power) / factorial(power)
        sign *= -1
    return result
def exp(x):
    result = 0
    for i in range(15):
        result += (x ** i) / factorial(i)
    return result
def f(x):
    return exp(sin(x))
pi = 3.141592653589793
n = int(input("Enter the number of intervals (even): "))
if n % 2 != 0:
    print("Error: Number of intervals must be even.")
else:
    a = 0
    b = pi / 2
    h = (b - a) / n
    total = f(a) + f(b)
    for i in range(1, n):
        x = a + i * h
        if i % 2 == 0:
            total += 2 * f(x)
        else:
            total += 4 * f(x)
    area = (h / 3) * total
    print("Approximate Area =", area)






# Simpson 3/8 rule

def factorial(n):
    fact = 1
    for i in range(1, n + 1):
        fact *= i
    return fact
def sin(x):
    result = 0
    sign = 1
    for i in range(10):
        power = 2 * i + 1
        result += sign * (x ** power) / factorial(power)
        sign *= -1
    return result
def exp(x):
    result = 0
    for i in range(15):
        result += (x ** i) / factorial(i)
    return result
def f(x):
    return exp(sin(x))
pi = 3.141592653589793
n = int(input())
if n % 3 != 0:
    print("Error")
else:
    a = 0
    b = pi / 2
    h = (b - a) / n
    total = f(a) + f(b)
    for i in range(1, n):
        x = a + i * h
        if i % 3 == 0:
            total += 2 * f(x)
        else:
            total += 3 * f(x)
    area = (3 * h / 8) * total
    print("Approximate Area =", area)




# Euler's Method

def f(x, y):
    return x**3 + y
x = float(input())
y = float(input())
h = float(input())
xn = float(input())
while x < xn:
    y = y + h * f(x, y)
    x = x + h
print(y)



# Runge Kutta

def f(x, y):
    return x + y
x = float(input())
y = float(input())
h = float(input())
xn = float(input())
while x < xn:
    k1 = h * f(x, y)
    k2 = h * f(x + h/2, y + k1/2)
    k3 = h * f(x + h/2, y + k2/2)
    k4 = h * f(x + h, y + k3)
    y = y + (k1 + 2*k2 + 2*k3 + k4) / 6
    x = x + h
    print(x, y)
```
